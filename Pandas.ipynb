# Series & DataFrame Basics

```python
# series (1D, labeled)
S1 = pd.Series([65, 12, 90],index = ["a", "b", "c"])
print(S1["c"])
print(S1.values)
print(S1.index)
```

```python
# DataFramea (2D Table)
df1 = pd.DataFrame({"City":["Chennai", "Bomaby","Bengaluru","Pune"],
                    "Name":["Thomas", "Andrew","William","Josh"],
                    "Age":[24,31,29,23]}, index = [1,2,3,4]) #making DataFrames from dictionary
print(df1)
```

```python
# surveying the data
df1.head() #by default first 5 rows if nothing given
df1.tail(2) #by default last 5 rows if nothing given
df1.shape #gives dimensions of DataFrame (rows, columns)
df1.columns #gives column names
df1.index #returns row labels
df1.dtypes #gives datatype of each column
```

```python
# to summarize a DataFrame
df1.info() #gives information of each column.
df1.describe() #gives statistical details of DataFrame
df1.describe(include = "object") #for text columns
```

```python
# one column of a DataFrame is a Series
df1["Age"]
type(df1["Age"]) # returns Pandas Series Datatype
df1["Age"].mean()
```

```python
# connection with numpy
df1["Age"].to_numpy() #this series is converted to an numpy array now
pd.DataFrame(np.arange(12).reshape(3,4)) #here numpy array is converted to DataFrame
```

# Selecting Data

```python
df = pd.DataFrame({
    "name":["Ram", "Prem", "Ujjwal", "Praful"],
    "age":[26, 30, 29, 25],
    "City":["Chennai", "Mumbai", "Delhi", "Bengaluru"]
})
df
```

```python
#selecting columns
print(df["age"])
print(df[["name","age"]])
```

```python
# using iloc (here end is excluded) and here we use position
df.iloc[0] #row of index zero
df.iloc[0:2] #row of index 0 and 1 here 2 is excluded
df.iloc[:,0:2] # all rows and columns of index 0 and 1 (2 excluded)
df.iloc[[0,2],[0,1]] # row of index 0 and 2, column of index 0 and 1
```

```python
# using loc (here end is also included) and here we can use labels only for columns
df.loc[0]
df.loc[0:2]
df.loc[0:2, "name":"age"]
df.loc[[0,3],["name","City"]]
```

```python
df1 = pd.DataFrame({
    "name":  ["Amit", "Riya", "Kabir", "Sara", "Neha"],
    "age":   [21, 22, 20, 23, 22],
    "city":  ["Delhi", "Pune", "Delhi", "Agra", "Pune"],
    "marks": [78.5, 88.0, 65.0, 92.5, 71.0]
})
```

```python
# Filtering (boolean masking, just like NumPy)
df1[df1["marks"] > 75]
df1[df1["city"] == "Delhi"]
df1[(df1["age"] >= 22) & (df1["marks"] > 70)] 
```

```python
# isin, between, string matchings
df1[df1["city"].isin(["Delhi","Agra"])]
df1[df1["marks"].between(60,75)]
df1[df1["name"].str.contains('a')]
df1[df1["name"].str.startswith('S')]
```

```python
# Filter + column together
df1.loc[df1["marks"]>70,["name","marks"]]
```

```python
# changing the index
df2 = df1.set_index("name")
df2.loc["Riya"]
df2.loc["Neha", "age"]
df2.loc["Kabir", ["city","marks"]]
df2.reset_index() #shows the older indexed
```

```python
# to modify values
df1.loc[df1["marks"] < 70, "marks"] = 70
```

# NumPy :- NumPy is a python library which allows us to make multi-dimensional arrays(vectors, matrices) and perform vectorized mathematics operations on those, these are faster than python lists and form a basis for data science.

```python
#let us import numpy for our future ease of purpose
import numpy as np
```

# Basics of creating NumPy arrays and modifying them

```python
a = np.array([4, 6, 2, 7, 7, 8]) # making 1-D array
print(a)
b = np.array([[4, 8, 2, 9],[1, 5, 0, 3]]) # making 2-D array
print(b)
```

```python
# some readymade numpy arrays
c = np.zeros((2,2))
print(c)
```

```python
c = np.ones((3,4))
print(c)
```

```python
d = np.full((2,3),7) #keeps all 7 (or desired values) in required dimension array
print(d)
```

```python
# to make identity matrix of required size
a = np.eye(5)
print(a)
```

```python
# np.arange(start, end, skip_value) - end not included
a = np.arange(0,10,2)
print(a)
```

```python
# np.linspace(start, end, no_of_elements) - here end is also included
b = np.linspace(0, 1, 5)
print(b)
```

```python
# to see properties of a numpy array
b = np.array([[4,6,7,8],[1,9,2,6],[8,3,9,2]])
print(np.shape(b))
print(np.ndim(b))
print(np.size(b))
print(b.dtype)
```

```python
# to set/update datatype of elements
b = np.array([4,7,2,9],dtype = float) # to set datatype of elements in an array
c = b.astype(int) # to change datatype of an array
print(c)
```

# Indexing and Slicing

```python
# for 1-D array indexing and slicing
a = np.array([6, 8, 12, 9, 10])
print(a[0])
print(a[1:4])
print(a[-3])
print(a[::2]) # traverse a numpy array with a give skip/gap value
print(a[::-1]) # reverse of a numpy array
```

```python
# for 2-D array indexing and slicing
m = np.array([[9, 2, 1, 6],
              [8, 7, 4, 3],
              [2, 6, 9, 0]])
# here the format is like m[row, column] - remembering this will make things most easy.
print(m[1,2])
print(m[0]) # gives first row of the matrix
print(m[:,1]) # gives second column of the matrix
print(m[1: , 1:3]) # generalised example
```

```python
# boolean masking
a = np.array([10, 20, 30, 40, 50, 60, 70])
print(a>25) 
print(a[(a < 50) & (a > 15)]) # here use & not 'and'
```

```python
#modifying values
a[(a > 15) & (a < 50)] = 14
print(a)
```

```python
# creating a new array from another array
a = [23, 65, 87, 12, 90, 43]
s = a[1:4]
s[0] = 39
print(s)
```

# Operations & Broadcasting

```python
a = np.array([1,2,3])
b = np.array([20,30,40])
```

```python
# corresponding element wise operations between two matrices
print(a+b)
print(a*b) # this is not the mathematical matrix multiplication
print(b/a)
```

```python
# operation for each element of an array with a scaler
print(a**2)
print(a + 5)
print(a*2)
```

```python
# some universal functions
print(np.sqrt(a))
print(np.sin(a))
print(np.cos(a))
print(np.log(a))
print(np.exp(a))
print(np.abs(np.array([-5, 8, -9, 0, -8, -2])))
```

```python
# comparisons
a = np.array([6, 7, 8])
b = np.array([69,83,47])
print(np.array_equal(a,b))
```

```python
# broadcasting
m = np.array([[1, 2, 3],
              [4, 5, 6]]) 

m + np.array([[100], [200]]) 
```

```python
m + np.array([100, 200, 300])
```

# Aggregations

```python
# axis = 0 - column   |   axis = 1 - row
```

```python
m = np.array(
    [[2.9, 5.6, 8.1],
    [3.7, 7.1,5.9],
    [2.7, 6.8, 3.5]]
)
# operations on complete array 
print(m.sum())
print(m.prod()) # product of all elements
print(m.mean()) # average
print(m.min())
print(m.max())
print(m.std()) # standard deviation
print(m.var()) # variance
```

```python
# if we work with axis
print(m.sum(axis = 1)) #working row wise
print(m.sum(axis = 0)) #working column wise
```

```python
# finding indexes
a = np.array([32, 65, 12, 90, 4])
print(np.argmax(a)) #argument of largest element
print(np.argmin(a)) #argument of smallest element
print(np.argsort(a)) #presenting indices in order of element sorting
```

```python
# sorting
print(np.sort(a))
m = np.array([[5, 7, 1, 9],
              [3, 8, 1, 5],
              [3,21,5,9]])
print(np.sort(m, axis = 0)) #sorted colummns
print(np.sort(m, axis = 1)) #sorted rows
```

```python
#cumulative functions
b = np.array([30, 17, 23, 50])
print(np.cumsum(b)) #cumulative sum
print(np.cumprod(b)) #cumulative product
```

```python
# conditional aggregates
a = np.array([21, 4, 9, 3, 16, 18, 12])
print(((a > 10) & (a < 20)).sum()) # gives count not actual sum

# the below two will give boolean results
print(np.any(a > 5))
print(np.all(a > 5))
```

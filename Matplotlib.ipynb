# Box Plot
Let us understand the terminologies behind this first. Suppose we have a sorted data given below. 
10, 12, 14, 16, 18, 20, 22, 24, 26

# Median (Q2)
Middle value of the sorted data. This divides data into two halves. For the above data median is 18.
# First Quartile (Q1)
Median value of the first half of sorted data. 25% percent of data lies below this Value. Here Q1 is 13.
# Third Quartile (Q3)
Median value of the second half of the data. 75% of data lies below this value. Here Q3 is 23.
# Interquartile Range (IQR)
Measures spread of 50% data. IQR = Q3 - Q1. Here IQR = 23 - 13 = 10.
# Whiskers
Lines that go outside the box. They show the highest and lowest values observations and help to decide boundaries to differentiate outliers from inliers.
Lower boundary = (Q1-1.5*(IQR))
Upper boundary = (Q3+1.5*(IQR))

# Bar Chart

```python
Company = ["Apple", "Microsoft", "Meta", "Starlink"]
Revenue = [58, 67, 59, 60]
plt.bar(Company, Revenue, color = ['red','blue','green','orange'])

plt.xlabel("Company Names")
plt.ylabel("Company Revenues (In Billion Dollars)")
plt.title("Revenue Comparison of Tech Giants")
plt.show()
```

```python
# horizontal bar graph
plt.barh(Company, Revenue, color = ['red','blue','green','orange'])

plt.xlabel("Company Revenues (In Billion Dollars)")
plt.ylabel("Company Names")
plt.title("Revenue Comparison of Tech Giants")
plt.show()
```

```python
# example of double bar graph
quarters = ["Q1", "Q2", "Q3", "Q4"]
team_A = [20, 35, 30, 40]
team_B = [25, 30, 35, 45]

x = np.arange(len(quarters))
width = 0.40

plt.bar(x - width/2, team_A, width, label="Team A")
plt.bar(x + width/2, team_B, width, label="Team B")

plt.xticks(x, quarters)
plt.xlabel("Quarter")
plt.ylabel("Sales")
plt.title("Quarterly Sales Comparison")
plt.legend()
plt.show()
```

# Line Charts

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [280, 230, 190, 250, 240, 260]

plt.plot(months, sales)
plt.title("Months v/s Sales Trend")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.show()
```

```python
# more customization to our line chart

plt.plot(
    months, sales,
    color="red",
    marker="o",
    linestyle="--",
    linewidth=2,
    label="Sales"
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)
plt.show()
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
team_A = [100, 120, 90, 150, 180]
team_B = [80, 110, 130, 140, 160]

plt.plot(months, team_A, marker="o", label="Team A")
plt.plot(months, team_B, marker="s", label="Team B") # s means square here.

plt.title("Monthly Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)

plt.show()
```

# Pie Chart

```python
# We know what a pie chart is let us directly start with the code
products = ["Mobile", "Laptop", "Tablet"]
Revenue = [40, 15, 45]

plt.pie(
        Revenue,
        labels = products,
        autopct = "%1.1f%%",   #1.1f --> means that minimum once character before the decimal and only one character after decimal
        startangle = 90        # startangle = 90 --> sets 90° as the startangle of pie chart
)

plt.title("Revenue of different products in electronics market")
plt.show()
```

# Density Plot
This is meant to draw probability density graph of a continuous random variable.

```python
marks = [45, 50, 55, 60, 62, 65, 68, 70, 72, 75, 78, 80, 82, 85, 90, 92]

plt.hist(marks, bins = 6, density = True, edgecolor = "black", color = "red") #in this histogram we are using density instead of frequency

plt.xlabel("Marks")
plt.ylabel("Density")
plt.title("Density Histogram")

plt.show()
```

# Violin Plot
Shows distribution as well as density for differenet random variables. Thus comparing multiple random variables is easier here.

```python
team_A = [45, 50, 52, 55, 56, 58, 60, 62, 65, 90]
team_B = [55, 60, 62, 63, 65, 66, 68, 70, 72, 75]

plt.violinplot([team_A, team_B], showmedians=True)

plt.xticks([1, 2], ["Team A", "Team B"])
plt.ylabel("Scores")
plt.title("Score Distribution by Team")

plt.show()
```

# Error Bar Graph
Error Graphs are used to find uncertainity or variability of a measurement.

```python
from IPython.display import Image
Image(r"C:\Users\Rayan chakraborty\OneDrive\画像\Screenshots\Screenshot 2026-10-22 151140.png", width=500)
```

```python
months = ["Jan", "Feb", "Mar"]
avg_delivery = [3, 5, 4]
error = [0.5, 1.0, 0.7]

plt.errorbar(
    months,
    avg_delivery,
    yerr=error,  # makes the error bars vertical in nature
    fmt="o",     # marker of value about which error is measured
    capsize=5    # size of caps at extremums of bars
)

plt.xlabel("Month")
plt.ylabel("Average Delivery Time (days)")
plt.title("Delivery Time with Error Bars")
plt.grid(True, axis="y")
plt.show()
```

# Subplots
To show multiple graphs on same canvas

```python
months = ["Jan", "Feb", "Mar", "Apr"]
sales = [20, 35, 30, 45]
profit = [5, 10, 8, 15]

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# First graph: Sales
axes[0].plot(months, sales, marker="o")
axes[0].set_title("Monthly Sales")
axes[0].set_xlabel("Months")
axes[0].set_ylabel("Sales")

# Second graph: Profit
axes[1].bar(months, profit)
axes[1].set_title("Monthly Profit")
axes[1].set_xlabel("Months")
axes[1].set_ylabel("Profit")

plt.tight_layout()
plt.show()
```

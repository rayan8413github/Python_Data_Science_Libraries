```python
x = [1, 2, 3, 4, 5, 6, 7]
y = [21, 90, 34, 46, 12, 10, 94]

plt.plot(x,y)
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.title("Heading of this graph")
plt.grid(True) # this particular line of code is responsible to show grid lines of a graph which helps in tracing in a better way 

plt.show()
```

```python
# let us understand plt.legend() now

months = [1, 2, 3, 4, 5, 6]
product_A_sales = [34, 56, 21, 78, 19, 47]
product_B_sales = [54, 26, 21, 87, 90, 40]

plt.plot(months,product_A_sales,label = "Product A")
plt.plot(months,product_B_sales,label = "Product B")

plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

# Histogram plotting

```python
marks = [35, 40, 42, 45, 48, 52, 55, 60, 75, 90]

plt.hist(marks, bins=5, edgecolor="black") #bins is the number of intervals we have and at y axis we have frequency axis

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

# Scatter Plot - It is very useful to relationship between two numerical variables

```python
study_hours = [1, 2, 3, 4, 5, 6, 7, 8]
marks = [43, 43.5, 43.5, 44, 44.5, 46, 47.5, 49]

plt.scatter(study_hours, marks)
plt.title("Study Hours vs Exam Marks")
plt.xlabel("Study Hours")
plt.ylabel("Exam Marks")

plt.show()
```

# For scatter plot remember the following things - 
 1) Positive Relationship - when x increases, y also increases.
 2) Negative Relationship - When x increases, y decreases.
 3) No Clear Relationship - No proper observation can be made to find a relationship between these two.

# Box Plot
Let us understand the terminologies behind this first. Suppose we have a sorted data given below. 
10, 12, 14, 16, 18, 20, 22, 24, 26

# Median (Q2)
Middle value of the sorted data. This divides data into two halves. For the above data median is 18.
# First Quartile (Q1)
Median value of the first half of sorted data. 25% percent of data lies below this Value. Here Q1 is 13.
# Third Quartile (Q3)
Median value of the second half of the data. 75% of data lies below this value. Here Q3 is 23.
# Interquartile Range (IQR)
Measures spread of 50% data. IQR = Q3 - Q1. Here IQR = 23 - 13 = 10.
# Whiskers
Lines that go outside the box. They show the highest and lowest values observations and help to decide boundaries to differentiate outliers from inliers.
Lower boundary = (Q1-1.5*(IQR))
Upper boundary = (Q3+1.5*(IQR))

```python
salary = [20, 22, 24, 25, 26, 28, 30, 32, 35, 90] #here 90 can be a potential outlier
plt.boxplot(salary)
plt.title("Salary Distribution")
plt.ylabel("Salary (In Thousands)")
plt.show()
```

```python
# compares salary distribution of two teams

salary_A = [20, 22, 24, 25, 26, 28, 30]
salary_B = [25, 30, 32, 35, 40, 45, 80]

plt.boxplot([salary_A, salary_B],
            tick_labels=["Team A", "Team B"])

plt.ylabel("Salary (in thousands)")
plt.title("Salary Comparison")
plt.show()
```

```python
# by default this graph is always vertical in shape but we can also show this graph in horizontal orientation

salary = [20, 22, 24, 25, 26, 28, 30, 32, 35, 90]
plt.boxplot(salary, orientation = 'horizontal')
plt.title("Salary Distribution")
plt.ylabel("Salary (In Thousands)")
plt.show()
```
